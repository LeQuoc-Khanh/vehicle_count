```python
import cv2
import os


input_folder = r"C:\Users\DELL\Documents\train\images"
output_folder = r"C:\Users\DELL\Documents\anh_da_xu_ly"

os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.endswith(('.jpg', '.png')):
        img = cv2.imread(os.path.join(input_folder, filename))
        
        # Resize về 640x640
        resized = cv2.resize(img, (640, 640))
        
        # Giảm nhiễu Gaussian
        denoised = cv2.GaussianBlur(resized, (3, 3), 0)
        

        cv2.imwrite(os.path.join(output_folder, filename), denoised)
        print(f"Đã xử lý: {filename}")

print("✅ Xử lý xong tất cả ảnh!")
```
